```python
# %%
import networkx as nx

from skimage.morphology import skeletonize


def skeleton_graph(skeleton):
    G = nx.Graph()
    pixels = set(
        map(
            tuple,
            np.argwhere(skeleton),
        )
    )

    for p in pixels:
        G.add_node(p)

        for dr in (-1, 0, 1):
            for dc in (-1, 0, 1):
                q = (
                    p[0] + dr,
                    p[1] + dc,
                )

                if q in pixels and q != p:
                    G.add_edge(
                        p,
                        q,
                        weight=np.hypot(
                            dr,
                            dc,
                        ),
                    )

    return G
```

```python
# %%
def prune_branches(
    G,
    min_length=20,
):
    G = G.copy()

    while True:
        endpoints = [
            n
            for n in G
            if G.degree[n] == 1
        ]

        removed = False

        for endpoint in endpoints:
            path = [
                endpoint
            ]

            length = 0
            previous = None
            current = endpoint

            while (
                G.degree[current] == 1
                or G.degree[current] == 2
            ):
                neighbours = [
                    n
                    for n in G.neighbors(
                        current
                    )
                    if n != previous
                ]

                if not neighbours:
                    break

                nxt = neighbours[0]

                length += G[
                    current
                ][nxt]["weight"]

                path.append(nxt)

                previous, current = (
                    current,
                    nxt,
                )

                if G.degree[current] != 2:
                    break

            if length < min_length:
                G.remove_nodes_from(
                    path[:-1]
                )

                removed = True
                break

        if not removed:
            break

    return G
```

```python
# %%
def tree_metrics(G):
    tree_length = G.size(
        weight="weight"
    )

    endpoints = [
        n
        for n in G
        if G.degree[n] == 1
    ]

    diameter_length = -1
    diameter_path = None

    for start in endpoints:
        lengths, paths = (
            nx.single_source_dijkstra(
                G,
                start,
                weight="weight",
            )
        )

        for end in endpoints:
            if (
                lengths[end]
                > diameter_length
            ):
                diameter_length = lengths[
                    end
                ]

                diameter_path = paths[
                    end
                ]

    traversal_length = (
        2 * tree_length
        - diameter_length
    )

    return (
        tree_length,
        diameter_length,
        traversal_length,
        diameter_path,
    )
```

```python
from IPython.display import display
def branching_skeleton_length(
    mask,
    min_branch_length=20,
):
    skeleton = skeletonize(
        mask
    )

    G = skeleton_graph(
        skeleton
    )

    G = prune_branches(
        G,
        min_length=min_branch_length,
    )

    (
        tree_length,
        diameter,
        traversal,
        diameter_path,
    ) = tree_metrics(
        G
    )

    tree = np.zeros_like(
        mask,
        dtype=bool,
    )

    for r, c in G.nodes:
        tree[r, c] = True

    return {
        "skeleton": skeleton,
        "tree": tree,
        "graph": G,
        "tree_length": tree_length,
        "diameter": diameter,
        "traversal_length": traversal,
        "diameter_path": diameter_path,
    }
```

```python
# %%
object_results = []
object_data = {}

height = int(target_image["height"])
width = int(target_image["width"])

for object_index, annotation in enumerate(
    target_annotations
):
    annotation_id = int(
        annotation["id"]
    )

    mask = decode_annotation_mask(
        annotation,
        height,
        width,
    ).astype(bool)

    result = branching_skeleton_length(
        mask,
        min_branch_length=20,
    )

    object_results.append(
        {
            "object_index": object_index,
            "annotation_id": annotation_id,
            "mask_area_px": int(mask.sum()),
            "tree_length_px": (
                result["tree_length"]
            ),
            "tree_length_um": (
                result["tree_length"]
                * MICRONS_PER_PIXEL
            ),
            "diameter_px": (
                result["diameter"]
            ),
            "diameter_um": (
                result["diameter"]
                * MICRONS_PER_PIXEL
            ),
            "traversal_length_px": (
                result["traversal_length"]
            ),
            "traversal_length_um": (
                result["traversal_length"]
                * MICRONS_PER_PIXEL
            ),
        }
    )

    object_data[annotation_id] = {
        "mask": mask,
        **result,
    }

length_results = pd.DataFrame(
    object_results
)

display(length_results)
```

```python
# %%
OBJECTS_PER_PAGE = 20
NUMBER_OF_COLUMNS = 4
CROP_PADDING = 6

ordered_results = (
    length_results
    .sort_values("object_index")
    .reset_index(drop=True)
)

number_of_pages = int(
    np.ceil(
        len(ordered_results)
        / OBJECTS_PER_PAGE
    )
)

for page_index in range(number_of_pages):
    page_results = ordered_results.iloc[
        page_index * OBJECTS_PER_PAGE:
        (page_index + 1) * OBJECTS_PER_PAGE
    ]

    number_of_rows = int(
        np.ceil(
            len(page_results)
            / NUMBER_OF_COLUMNS
        )
    )

    figure, axes = plt.subplots(
        number_of_rows,
        NUMBER_OF_COLUMNS,
        figsize=(
            14,
            3.7 * number_of_rows,
        ),
        squeeze=False,
    )

    axes = axes.ravel()

    for axis, row in zip(
        axes,
        page_results.itertuples(
            index=False
        ),
    ):
        annotation_id = int(
            row.annotation_id
        )

        data = object_data[
            annotation_id
        ]

        mask = data["mask"]
        skeleton = data["skeleton"]
        tree = data["tree"]
        diameter_path = data[
            "diameter_path"
        ]

        mask_rows, mask_columns = np.nonzero(
            mask
        )

        row_start = max(
            int(mask_rows.min())
            - CROP_PADDING,
            0,
        )
        row_stop = min(
            int(mask_rows.max())
            + CROP_PADDING
            + 1,
            mask.shape[0],
        )
        column_start = max(
            int(mask_columns.min())
            - CROP_PADDING,
            0,
        )
        column_stop = min(
            int(mask_columns.max())
            + CROP_PADDING
            + 1,
            mask.shape[1],
        )

        crop = (
            slice(row_start, row_stop),
            slice(column_start, column_stop),
        )

        mask_crop = mask[crop]
        skeleton_crop = skeleton[crop]
        tree_crop = tree[crop]

        overlay_rgb = np.zeros(
            (*mask_crop.shape, 3),
            dtype=float,
        )

        overlay_rgb[mask_crop] = [
            0.40,
            0.40,
            0.40,
        ]

        overlay_rgb[skeleton_crop] = [
            1.0,
            0.45,
            0.0,
        ]

        overlay_rgb[tree_crop] = [
            0.0,
            1.0,
            1.0,
        ]

        axis.imshow(
            overlay_rgb,
            interpolation="nearest",
        )

        if diameter_path:
            diameter_rows = [
                point[0] - row_start
                for point in diameter_path
            ]

            diameter_columns = [
                point[1] - column_start
                for point in diameter_path
            ]

            axis.plot(
                diameter_columns,
                diameter_rows,
                color="magenta",
                linewidth=2.0,
            )

        axis.set_title(
            f"Object {int(row.object_index)} | "
            f"ann. {annotation_id}\n"
            f"tree={row.tree_length_px:.1f} px | "
            f"diameter={row.diameter_px:.1f} px\n"
            f"traversal="
            f"{row.traversal_length_px:.1f} px",
            fontsize=8,
        )

        axis.axis("off")

    for axis in axes[
        len(page_results):
    ]:
        axis.axis("off")

    figure.suptitle(
        f"{TARGET_FILENAME}\n"
        f"page {page_index + 1}/{number_of_pages}\n"
        "mask: gray | skeleton: orange | "
        "pruned tree: cyan | diameter: magenta",
        fontsize=12,
    )

    figure.tight_layout()
    plt.show()
```
